```python
class Student:
    def __init__(self, r, n, c, e):
        self.roll_no = r
        self.name = n
        self.city = c
        self.email = e


s1 = Student(1, "Jay", "Pune", "jay@gmail.com")

print(s1.name)
print(s1.city)
```

```python
class Laptop:
    company = "Lenovo"

    def __init__(self, c, r):
        self.color = c
        self.RAM = r


l1 = Laptop("Black", "8GB")

print(l1.company)
print(l1.color)
print(l1.RAM)
```
